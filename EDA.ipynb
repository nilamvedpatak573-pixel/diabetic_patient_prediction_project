```python
# import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
# import dataset
df = pd.read_csv('cleaned_data.csv')
```

```python
df.info()
```

```python
# pregnancies bar plot
ax = df['Pregnancies'].value_counts().sort_index().plot(kind = 'bar')
ax.bar_label(ax.containers[0])
plt.show()
```

```python
# glucose ppregnancies bar plot
for column in['Glucose','BloodPressure','SkinThickness','Insulin','BMI','DiabetesPedigreeFunction','Age']:
    print(column)
    ax = df[column].plot(kind = 'hist')
    plt.show()
```
